```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

torch.manual_seed(1)
```

```python
# y=2x

# 데이터
x_train = torch.FloatTensor([[1], [2], [3]])
y_train = torch.FloatTensor([[2], [4], [6]])
```

```python
# 모델 선언 및 초기화.
# 하나의 입력 x에 대해서 하나의 출력 y를 갖음
# input_dim = 1 / output_dim = 1
model1 = nn.Linear(1, 1)

print(list(model1.parameters()))
```

```python
optimizer = optim.SGD(model1.parameters(), lr=0.01)

nb_epochs = 2000
for epoch in range(nb_epochs + 1):

    # H(x) 계산
    prediction = model1(x_train)

    # cost 계산
    cost = F.mse_loss(prediction, y_train)

    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    if epoch % 100 == 0:
    # 100번마다 로그 출력
      print('Epoch {:4d}/{} Cost: {:.6f}'.format(
          epoch, nb_epochs, cost.item()
      ))
```

```python
# 임의의 입력 4 선언
new_var = torch.FloatTensor([[4.0]])
pred_y = model1(new_var)
print(pred_y)
```

```python
# H(x) = w1x1 + w2x2 + w3x3 + b
x_train = torch.FloatTensor([[73, 80, 75],
                             [93, 88, 93],
                             [89, 91, 90],
                             [96, 98, 100],
                             [73, 66, 70]])
y_train = torch.FloatTensor([[152], [185], [180], [196], [142]])
```

```python
# 모델 선언
model2 = nn.Linear(3, 1)
print(list(model2.parameters()))
```

```python
optimizer = optim.SGD(model2.parameters(), lr=1e-5)

nb_epochs = 2000
for epoch in range(nb_epochs + 1):

    # H(x)
    prediction = model2(x_train)

    # cost 계산
    cost = F.mse_loss(prediction, y_train)

    optimizer.zero_grad()
    cost.backward()
    optimizer.step()

    if epoch % 100 == 0:
    # 100번마다 로그 출력
      print('Epoch {:4d}/{} Cost: {:.6f}'.format(
          epoch, nb_epochs, cost.item()
      ))
```

```python
# 임의의 입력 [73, 80, 75]를 선언
new_var =  torch.FloatTensor([[73, 80, 75]])
pred_y = model2(new_var)
print(pred_y)
```

```python
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.linear(x)

model3 = LinearRegressionModel()
```

```python
class MultivariateLinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(3, 1)

    def forward(self, x):
        return self.linear(x)

model4 = MultivariateLinearRegressionModel()
```
